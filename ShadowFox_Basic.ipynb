## Visualization Library Documentation
### MATPLOTLIB

```python
import matplotlib.pyplot as plt ## Import the library
import numpy as np
```

```python
## BAR CHART
categories = ['A','B','C','D','E']
values = [10,20,15,6,13]
plt.figure(figsize=(5,5))         # width=5, height=5
plt.bar(categories, values, color='orange')
plt.title("Bar Chart")
plt.show()
```

```python
## LINE CHART
x = [2020, 2021, 2022, 2023, 2024]   # years
y = [100, 150, 180, 130, 150]        # profit
plt.figure(figsize=(5,4))
plt.plot(x, y, color='blue', marker='o')
plt.title(" Profit Analysis")
plt.xlabel("Year")
plt.ylabel("Profit")

plt.xticks(x)                      # Show only year values on x-axis

plt.show()
```

```python
## Scattered plot

x = [1,2,3,4,5,6,7,8,9]
y = [5,7,8,7,6,10,5,3,2]

plt.scatter(x,y, color='red', marker='*')
plt.title("Scatter Plot")
plt.show()
```

```python
data = np.random.randn(1000) # 1000 data points from a normal distribution

plt.hist(data, bins=30, color='lightgreen', edgecolor='black')
plt.title("Distribution of Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## SEABORN

```python
import seaborn as sns
import plotly.express as px
```

```python
import pandas as pd
```

```python
tips = sns.load_dataset("tips")  ## loading data table "tips"
```

```python
##scatterplot
sns.scatterplot(x="total_bill", y="tip", hue="sex", data=tips)
plt.show()
```

```python
## box plot
sns.boxplot(x="sex", y="tip", data=tips)
plt.show()
```

```python
## count plot
sns.countplot(x="smoker", data=tips)
plt.show()
```

```python
## Heat map
corr = tips.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
## Violin plot
sns.violinplot(x="day", y="total_bill", data=tips)
plt.show()
```

```python
sns.barplot(x="day", y="tip", hue="smoker", data=tips)
plt.show()
```
